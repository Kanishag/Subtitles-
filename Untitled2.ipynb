```python
import os, re, shutil, subprocess, tempfile, time
import torch, whisper

_models = {}
def get_model(name):
    if name not in _models:
        _models[name] = whisper.load_model(name)
    return _models[name]

def fmt_ts(t):
    ms = int(round(t * 1000))
    h, ms = divmod(ms, 3600000)
    m, ms = divmod(ms, 60000)
    s, ms = divmod(ms, 1000)
    return f"{h:02d}:{m:02d}:{s:02d},{ms:03d}"

def write_srt(segments, path):
    with open(path, "w", encoding="utf-8") as f:
        for i, seg in enumerate(segments, 1):
            f.write(f"{i}\n{fmt_ts(seg['start'])} --> {fmt_ts(seg['end'])}\n{seg['text'].strip()}\n\n")

def transcribe(video_path, model_name="small", task="transcribe", language=None):
    return get_model(model_name).transcribe(
        video_path, task=task, language=language,
        fp16=torch.cuda.is_available())

def burn_subtitles(video_path, srt_path, out_path, font_size=24):
    work = tempfile.mkdtemp()
    ext = os.path.splitext(video_path)[1] or ".mp4"
    shutil.copy(video_path, os.path.join(work, "input" + ext))
    shutil.copy(srt_path, os.path.join(work, "subs.srt"))
    cmd = ["ffmpeg", "-y", "-i", "input" + ext, "-vf",
           f"subtitles=subs.srt:force_style='FontSize={font_size},Outline=2'",
           "-c:a", "copy", "output.mp4"]
    subprocess.run(cmd, cwd=work, check=True, capture_output=True)
    shutil.copy(os.path.join(work, "output.mp4"), out_path)

def make_subtitled_video(video_path, model_name="small", task="transcribe",
                         language=None, font_size=24, out_dir="subtitled"):
    os.makedirs(out_dir, exist_ok=True)
    base = os.path.splitext(os.path.basename(video_path))[0]
    result = transcribe(video_path, model_name, task, language)
    srt_path = os.path.join(out_dir, base + ".srt")
    out_path = os.path.join(out_dir, base + "_subtitled.mp4")
    write_srt(result["segments"], srt_path)
    burn_subtitles(video_path, srt_path, out_path, font_size)
    return out_path, srt_path, result
```

```python
import pandas as pd, jiwer
reference_text = ""  # true transcript (optional)

def normalize(t):
    return re.sub(r"[^\w\s]", "", t.lower()).strip()

rows = []
for name in ["tiny", "base", "small"]:
    start = time.time()
    res = transcribe(video_file, name)
    row = {"model": name, "time_sec": round(time.time() - start, 1)}
    if reference_text.strip():
        row["WER"] = round(jiwer.wer(normalize(reference_text), normalize(res["text"])), 3)
    rows.append(row)
pd.DataFrame(rows)
```

```python
 import gradio as gr
LANGS = {"Auto-detect": None, "English": "en", "Tamil": "ta",
         "Hindi": "hi", "Telugu": "te", "Malayalam": "ml", "Kannada": "kn"}

def ui_run(video, model_name, language, task, font_size):
    if video is None:
        raise gr.Error("Please upload a video.")
    out_video, out_srt, _ = make_subtitled_video(
        video, model_name, task, LANGS[language], int(font_size))
    return out_video, out_srt

demo = gr.Interface(
    fn=ui_run,
    inputs=[gr.Video(label="Upload video"),
            gr.Dropdown(["tiny", "base", "small", "medium"], value="small", label="Model"),
            gr.Dropdown(list(LANGS.keys()), value="Auto-detect", label="Language"),
            gr.Radio(["transcribe", "translate"], value="transcribe", label="Task"),
            gr.Slider(12, 48, value=24, step=2, label="Font size")],
    outputs=[gr.Video(label="Subtitled video"), gr.File(label=".srt")],
    title="Auto Subtitle Generator")
demo.launch(share=True)
```
